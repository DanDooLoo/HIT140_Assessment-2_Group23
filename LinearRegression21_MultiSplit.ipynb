```python
"""
Linear Regression Analysis with Multiple Test-Train Splits
Calculates metrics for 90% test sizes
Compares against Baseline (predicting mean)
"""

import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import warnings
warnings.filterwarnings('ignore')

print("="*70)
print("ENHANCED LINEAR REGRESSION WITH 90/10, 80/20, 70/30 TEST SPLITS")
print("="*70)
```

```python
# Load datasets
full_data = pd.read_csv('fullv2.csv')
external_data = pd.read_csv('externalv2.csv')
```

```python
# Merge data
dataset = full_data.merge(external_data, left_on='match_id', right_on='match_id', how='inner')

```

```python
# Create response variable
data = dataset.copy()
data['goal_diff'] = data['homeScore'] - data['awayScore']
```

```python
# Select EXACTLY 8 explanatory variables (all pre-match available)
feature_names = [
    'home_Possession',        # 1. Home team ball possession (%)
    'away_Possession',        # 2. Away team ball possession (%)
    'home_xG',                # 3. Home expected goals
    'away_xG',                # 4. Away expected goals
    'betting_odds_home',      # 5. Pre-match betting odds for home
    'referee_strictness',     # 6. Referee strictness rating
    'travel_distance_km',     # 7. Away team travel distance (km)
    'social_sentiment_home'   # 8. Home team social sentiment
]

final_df = data[feature_names + ['goal_diff']].dropna()
X = final_df[feature_names].copy()
y = final_df['goal_diff'].copy()

print(f"\nTotal dataset: {len(X)} matches\n")
```

```python
# ============================================================================
# TEST MULTIPLE SPLITS: 70%, 80%, 90%
# ============================================================================

results_summary_Model_1 = []
results_summary_Model_2 = []
results_summary_Model_3 = []

# Build linear regression model
print('\n[STEP 3] Building Linear Regression Model')
print('-'*70)

# Scale features (StandardScaler)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print('Features standardized (mean=0, std=1)')

# Split data 70/30 train/test with random_state=42
X_train_1, X_test_1, y_train_1, y_test_1 = train_test_split(
    X_scaled, y, test_size=0.3, random_state=42
)

# Split data 80/20 train/test with random_state=42 | Baseline
X_train_2, X_test_2, y_train_2, y_test_2 = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# Split data 90/10 train/test with random_state=42
X_train_3, X_test_3, y_train_3, y_test_3 = train_test_split(
    X_scaled, y, test_size=0.1, random_state=42
)
```

```python
# ========================================================================
# MLP PERFORMANCE: Linear Regression Model 1 - 70/30 Test Split
# ========================================================================
print(f"\nMLP PERFORMANCE:")
print("-" * 70)

model = LinearRegression()
model.fit(X_train_1, y_train_1)

y_pred_test_1 = model.predict(X_test_1)

# Calculate metrics
mae_1 = mean_absolute_error(y_test_1, y_pred_test_1)
mse_1 = mean_squared_error(y_test_1, y_pred_test_1)
rmse_1 = np.sqrt(mse_1)
rmse_normalized_1 = rmse_1 / y_test_1.std()  # Normalized by test set std
r2_1 = r2_score(y_test_1, y_pred_test_1)

intercept = model.intercept_
coef = model.coef_  # Array of coefficients
print(f"Intercept: {intercept:.6f}")
print(f"Coefficients: {[f'{c:.6f}' for c in coef]}")
print(f"\nMAE: {mae_1:.6f}")
print(f"MSE: {mse_1:.6f}")
print(f"RMSE: {rmse_1:.6f}")
print(f"RMSE (Normalised): {rmse_normalized_1:.6f}")
print(f"R²: {r2_1:.6f}")
```

```python
# ========================================================================
# MLP PERFORMANCE: Linear Regression Model 2 - 80/20 Test Split
# ========================================================================
print(f"\nMLP PERFORMANCE:")
print("-" * 70)

model = LinearRegression()
model.fit(X_train_2, y_train_2)

y_pred_test_2 = model.predict(X_test_2)

# Calculate metrics
mae_2 = mean_absolute_error(y_test_2, y_pred_test_2)
mse_2 = mean_squared_error(y_test_2, y_pred_test_2)
rmse_2 = np.sqrt(mse_2)
rmse_normalized_2 = rmse_2 / y_test_2.std()  # Normalized by test set std
r2_2 = r2_score(y_test_2, y_pred_test_2)

intercept = model.intercept_
coef = model.coef_  # Array of coefficients
print(f"Intercept: {intercept:.6f}")
print(f"Coefficients: {[f'{c:.6f}' for c in coef]}")
print(f"\nMAE: {mae_2:.6f}")
print(f"MSE: {mse_2:.6f}")
print(f"RMSE: {rmse_2:.6f}")
print(f"RMSE (Normalised): {rmse_normalized_2:.6f}")
print(f"R²: {r2_2:.6f}")
```

```python
# ========================================================================
# MLP PERFORMANCE: Linear Regression Model 3 - 90/10 Test Split
# ========================================================================
print(f"\nMLP PERFORMANCE:")
print("-" * 70)

model = LinearRegression()
model.fit(X_train_3, y_train_3)

y_pred_test_3 = model.predict(X_test_3)

# Calculate metrics
mae_3 = mean_absolute_error(y_test_3, y_pred_test_3)
mse_3 = mean_squared_error(y_test_3, y_pred_test_3)
rmse_3 = np.sqrt(mse_3)
rmse_normalized_3 = rmse_3 / y_test_3.std()  # Normalized by test set std
r2_3 = r2_score(y_test_3, y_pred_test_3)

intercept = model.intercept_
coef = model.coef_  # Array of coefficients
print(f"Intercept: {intercept:.6f}")
print(f"Coefficients: {[f'{c:.6f}' for c in coef]}")
print(f"\nMAE: {mae_3:.6f}")
print(f"MSE: {mse_3:.6f}")
print(f"RMSE: {rmse_3:.6f}")
print(f"RMSE (Normalised): {rmse_normalized_3:.6f}")
print(f"R²: {r2_3:.6f}")
```

```python
# ========================================================================
# BASELINE PERFORMANCE: Predict Mean - 70/30 Test Split
# ========================================================================
print(f"\nBASELINE PERFORMANCE: Predicting Mean - 70/30 Test Split:")
print("-" * 70)

# Baseline: predict the mean of training set
y_pred_baseline_1 = np.full_like(y_test_1, y_train_1.mean())

mae_baseline_1 = mean_absolute_error(y_test_1, y_pred_baseline_1)
mse_baseline_1 = mean_squared_error(y_test_1, y_pred_baseline_1)
rmse_baseline_1 = np.sqrt(mse_baseline_1)
rmse_baseline_normalized_1 = rmse_baseline_1 / y_test_1.std()
r2_baseline_1 = r2_score(y_test_1, y_pred_baseline_1)

print(f"MAE: {mae_baseline_1:.6f}")
print(f"MSE: {mse_baseline_1:.6f}")
print(f"RMSE: {rmse_baseline_1:.6f}")
print(f"RMSE (Normalised): {rmse_baseline_normalized_1:.6f}")
print(f"R²: {r2_baseline_1:.6f}")
```
