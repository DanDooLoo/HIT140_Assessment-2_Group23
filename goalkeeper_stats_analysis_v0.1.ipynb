```python
# Mean
saves_mean_stat_xbar = stats.mean(goalkeeper["Goalkeeper_Saves"])
inside_penalty_mean_stat_xbar = stats.mean(goalkeeper["Goalkeeper_Actions_inside_PenaltyArea"])
outside_penalty_mean_stat_xbar = stats.mean(goalkeeper["Goalkeeper_Actions_outside_PenaltyArea"])

print("savemeanxbar: %.2f" % saves_mean_stat_xbar)
print("IPAMSXB: %.2f" % inside_penalty_mean_stat_xbar)
print("OPAMSXB: %.2f" % outside_penalty_mean_stat_xbar)
```

```python
# Median
saves_median_stat = stats.median(goalkeeper["Goalkeeper_Saves"])
ipa_median_stat = stats.median(goalkeeper["Goalkeeper_Actions_inside_PenaltyArea"])
opa_median_stat = stats.median(goalkeeper["Goalkeeper_Actions_outside_PenaltyArea"])
print("saves median stat: %.2f" % saves_median_stat)
print("ipa median stat: %.2f" % ipa_median_stat)
print("opa median stat %.2f" % opa_median_stat)
```

```python
# Mode
gk_save_mode_stat = stats.mode(goalkeeper["Goalkeeper_Saves"])
gk_ipa_mode_stat = stats.mode(goalkeeper["Goalkeeper_Actions_inside_PenaltyArea"])
gk_opa_mode_stat = stats.mode(goalkeeper["Goalkeeper_Actions_outside_PenaltyArea"])
print("gk_save_mode_stat: %.2f" % gk_save_mode_stat)
print("gk_ipa_mode_stat: %.2f" % gk_ipa_mode_stat)
print("gk_opa_mode_stat: %.2f" % gk_opa_mode_stat)
```

```python
# Variance
gk_save_var = np.array(goalkeeper["Goalkeeper_Saves"])
gk_saves_square = gk_save_var.var(ddof=1) # sample var - degree of freedon (ddof)
gk_save_sigma_square = gk_save_var.var() # popluation var - no ddof required

print("variance s square: ", gk_saves_square)
print("var sigma square: ", gk_save_sigma_square)
```

```python
# standard deviation
s = gk_save_var.std(ddof=1) # sample standard dev
sigma = gk_save_var.std()
print("standard deviation: %.2f" % s)
print("stad dev population: ", sigma)
```

```python
x = x.reshape(-1, 1)
y = y.reshape(-1, 1)
s = s.reshape(-1, 1)
```

```python
# Model 1 #
# Initialise a linear regression model (with intercept)
model = LinearRegression()

# Fit the linear regression model on the data
model.fit(x, y)

# Get the intercept and coefficient values
intercept = model.intercept_
coeff_intercept = model.coef_ 
print("Model with intercept:")
print("Intercept: ", intercept)
print("Coefficient: ", coeff_intercept)

# Visualise the regression line over all data points
x_line1 = s
y_line1 = x_line1 * coeff_intercept + intercept
```

```python
plt.scatter(goalkeeper["Goalkeeper_Actions_inside_PenaltyArea"], goalkeeper["Goalkeeper_Actions_outside_PenaltyArea"])
plt.title("GK_inside_PA vs. GK_outside_PA")
plt.xlabel("Goalkeeper_Actions_inside_PenaltyArea")
plt.ylabel("Goalkeeper_Actions_outside_PenaltyArea")
plt.plot(x_line1, y_line1, ":g", label="y=1.0057 + 0.8276*x")
plt.legend(loc='upper left')
plt.grid()
plt.show()
```

```python
model2 = LinearRegression(fit_intercept=False)

# Fit the linear regression model on the data
model2.fit(x, y)

# Print the intercept and coefficient learned by the linear regression model
intercept_zero = 0
coeff_nointercept = model2.coef_ 
print("Model without intercept:")
print("Intercept: ", intercept_zero)
print("Coefficient: ", coeff_nointercept)

# Visualise the 2nd regression line over all data points
# Include the first regression line for easy comparison.
x_line2 = np.linspace(goalkeeper["Goalkeeper_Actions_inside_PenaltyArea"].min(), goalkeeper["Goalkeeper_Actions_inside_PenaltyArea"].max(), 100)
y_line2 = x_line2 * coeff_nointercept + intercept_zero

plt.scatter(goalkeeper["Goalkeeper_Actions_inside_PenaltyArea"], goalkeeper["Goalkeeper_Actions_outside_PenaltyArea"], color='blue', label='Data points')
plt.title("Total GK inside PA vs. Total GK outside PA")
plt.xlabel("Total GK inside PA")
plt.ylabel("Total GK outside PA")
plt.plot(x_line1, y_line1, ":g", label="y=1.0057 + 0.8276*x")
plt.plot(x_line2, y_line2, ":r", label="y=0.0 + 0.0950*x")
plt.legend(loc='upper left')
plt.grid()
plt.show()
```
